```python
import numpy as np

def integrand(t):
    # f(t) = sqrt(vx^2 + vy^2 + vz^2)
    return np.sqrt(2**2 + (2*t)**2 + np.cos(t)**2)

def regla_simpson(f, a, b, n):
    if n % 2 != 0:
        n += 1  # n debe ser par para la regla de Simpson
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = f(x)

    # Suma de Simpson
    integral = y[0] + y[-1] + 4 * np.sum(y[1:-1:2]) + 2 * np.sum(y[2:-2:2])
    integral *= h / 3
    return integral

# Intervalo [0, 2] con n = 1000 subdivisiones
a = 0.0
b = 2.0
n = 1000

longitud_arco = regla_simpson(integrand, a, b, n)
print(f"Longitud de arco aproximada (Desplazamiento de la trayectoria): {longitud_arco:.6f} metros")
```
